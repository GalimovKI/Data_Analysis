При выполнении каждой итерации с целью дальнейшего анализа сохраняйте текущее значение оптимизируемого функционала, а также затраченное время на итерацию.  **При реализации класса запрещено пользоваться ИИ-инструментами.**

*Замечания.*

2. Иногда при подсчете сигмоиды и оптимизируемого функционала могут возникать вычислительные ошибки. Для их избежания существуют специальные трюки.
    * [How to Evaluate the Logistic Loss and not NaN trying](http://fa.bianp.net/blog/2019/evaluate_logistic/)

4. Обратите внимание, что класс `LogisticRegression` &mdash; наследник класса `BaseEstimator`, это с легкостью позволит использовать наш класс в различных пайплайнах библиотеки `sklearn`.
4. Следите за качеством кода, комментируйте логические этапы кода. Несоблюдение этого требования может привести к потере баллов.


```python
# При реализации класса запрещено пользоваться ИИ-инструментами.


class LogisticRegression(BaseEstimator):
    """Модель логистической регрессии.

    Параметры:
    method (Literal['gd', 'sgd']): Метод оптимизации ('gd' - градиентный спуск, 
        'sgd' - стохастический градиентный спуск).
    learning_rate (float): Константа скорости обучения, на которую домножаем градиент при обучении
    tol (float): Допустимое изменение функционала между итерациями.
    max_iter (int): Максимальное число итераций.
    batch_size (int): Размер выборки для оценки градиента (используется только при 'sgd').
    fit_intercept (bool): Добавлять ли константу в признаки.
    save_history (bool): Сохранять ли историю обучения.
    """

    def __init__(
        self,
        method: Literal["gd", "sgd"] = "gd",
        learning_rate: float = 0.5,
        tol: float = 1e-3,
        max_iter: int = int(1e4),
        batch_size: int = 64,
        fit_intercept: bool = True,
        save_history: bool = True,
    ):
        """Создает модель и инициализирует параметры."""
        self.method = method
        self.learning_rate = learning_rate
        self.tol = tol
        self.max_iter = max_iter
        self.batch_size = batch_size
        self.fit_intercept = fit_intercept
        self.save_history = save_history
        self.history = []  # История обучения

    @staticmethod
    def _sigmoid(x: np.ndarray) -> np.ndarray:
        """Вычисляет сигмоидную функцию."""
        return 1 / (1 + np.exp(-x))

    def _add_intercept(self, X: np.ndarray) -> np.ndarray:
        """Добавляет свободный коэффициент к матрице признаков.

        Параметры: X (np.ndarray): Исходная матрица признаков.

        Возвращает: np.ndarray: Матрица X с добавленным свободным
        коэффициентом.
        """
        X_copy = np.full((X.shape[0], X.shape[1] + 1), fill_value=1)
        X_copy[:, :-1] = X
        return X_copy
    
    @staticmethod
    def _loss_function(X: np.ndarray, Y: np.ndarray, thetas: np.ndarray) -> float:
        Z = X @ thetas

        log_1 = -np.logaddexp(0, -Z).reshape(-1)
        log_0 = -np.logaddexp(0, Z).reshape(-1)
        loss = -np.mean(Y * log_1 + (1 - Y) * log_0)

        return loss

    def fit(self, X: np.ndarray, Y: np.ndarray) -> "LogisticRegression":
        """Обучает модель логистической регрессии.

        Также, в случае self.save_history=True, добавляет в self.history
        текущее значение оптимизируемого функционала и затраченное время.

        Параметры:
        X (np.ndarray): Матрица признаков.
        Y (np.ndarray): Вектор истинных меток.

        Возвращает:
        LogisticRegression: Обученная модель.
        """
        if X.shape[0] != Y.shape[0]:
            raise ValueError("Количество строк в X и Y должно совпадать")

        if self.fit_intercept:
            X_copy = self._add_intercept(X)
        else:
            X_copy = X.copy()
            
        Y = np.asarray(Y)

        #Задаем начальные значения
        n_samples, n_features = X_copy.shape
        thetas = np.zeros((n_features, 1)).reshape(-1)
        n_iter : int = 0

        if self.method == "gd" :
        #Часть с градиентным спуском
            start_time = time.time()
            for n_iter in range(self.max_iter) :
                #Считаем предыдущий функционал
                prev_functional = LogisticRegression._loss_function(X_copy, Y, thetas)

                #Обновляем thet'ы
                Z = X_copy @ thetas
                Y_hat = LogisticRegression._sigmoid(Z)
                grad_loss_func = (X_copy.T @ (Y_hat.reshape(-1) - Y)) / n_samples
                theta_delta = self.learning_rate * grad_loss_func
                thetas -= theta_delta

                #Считаем новую функцию потерь
                loss = LogisticRegression._loss_function(X_copy, Y, thetas)

                if self.save_history:
                    iter_time = time.time() - start_time
                    self.history.append((loss, iter_time))
                
                #Проверяем tol
                if abs(loss - prev_functional) <= self.tol:
                    break

                #Обновляем время
                start_time = time.time()
        else :
        #Часть с стохастическим градиентным спуском
            #Перемешаем данные
            indices = np.random.permutation(n_samples)
            X_shuffled = X_copy[indices]
            Y_shuffled = Y[indices]

            start_time = time.time()
            start_batch = 0

            for n_iter in range(self.max_iter) :
                #Создаем батч
                end_batch = min(n_samples, start_batch + self.batch_size)

                X_batch = X_shuffled[start_batch : end_batch]
                Y_batch = Y_shuffled[start_batch : end_batch]

                #Обновляем стартовую позицию нового батча
                start_batch = (start_batch + self.batch_size) % n_samples

                #Аналогично gd
                prev_functional = LogisticRegression._loss_function(X_shuffled, Y_shuffled, thetas)

                #Обновляем thet'ы
                Z_batch = X_batch @ thetas
                Y_batch_hat = LogisticRegression._sigmoid(Z_batch)
                grad_loss_func = (X_batch.T @ (Y_batch_hat.reshape(-1) - Y_batch)) / self.batch_size
                theta_delta = self.learning_rate * grad_loss_func
                thetas -= theta_delta                
                
                #Считаем новую функцию потерь
                loss = LogisticRegression._loss_function(X_shuffled, Y_shuffled, thetas)

                if self.save_history:
                    iter_time = time.time() - start_time
                    self.history.append((loss, iter_time))
                
                #Проверяем tol
                if abs(loss - prev_functional) <= self.tol:
                    break

                #Обновляем время
                start_time = time.time()

        self.coef_ = thetas[:-1] # Коэффициенты модели
        self.intercept_ = thetas[-1]  # Свободный коэффициент
        self.n_iter_ = n_iter + 1 # Число итераций

        return self

    def predict(self, X: np.ndarray) -> np.ndarray:
        """Возвращает предсказанные классы.

        Параметры: X (np.ndarray): Матрица признаков.

        Возвращает: np.ndarray: Предсказанные классы.
        """
        thetas = np.append(self.coef_, self .intercept_)
        n_samples = X.shape[0]
        
        if self.fit_intercept:
            X_copy = self._add_intercept(X)
        else:
            X_copy = X.copy()


        #Предсказанные вероятности класса 1
        predictions_proba = LogisticRegression._sigmoid(X_copy @ thetas)
        
        #Предсказанные классы
        predictions = np.array([0] * n_samples) + (predictions_proba > 0.5)

        return predictions

    def predict_proba(self, X: np.ndarray) -> np.ndarray:
        """Возвращает вероятности классов 0 и 1.

        Параметры: X (np.ndarray): Матрица признаков.

        Возвращает: np.ndarray: Матрица вероятностей классов (n_samples,
        2).
        """
        thetas = np.append(self.coef_, self .intercept_)


        if self.fit_intercept:
            X_copy = self._add_intercept(X)        
        else:
            X_copy = X.copy()
    
        #Предсказанные вероятности класса 1 и 0
        predictions_proba_1 = LogisticRegression._sigmoid(X_copy @ thetas)
        predictions_proba_0 = 1 - predictions_proba_1

        prob_predictions = np.column_stack((predictions_proba_0, predictions_proba_1))
        
        return prob_predictions
```

Напишем функцию для проведения эксперимента

```python
def run_experiment(X, Y, method, lr=0.5, mi=int(1e4)):
    model = LogisticRegression(method=method, learning_rate=lr, max_iter=mi)
    model.fit(X, Y)

    losses, times = zip(*model.history)
    iters = range(1, model.n_iter_ + 1)

    times = np.cumsum(times)
    
    return losses, times, iters

# Запуск для GD и SGD
gd_losses, gd_times, gd_iters = run_experiment(X_train_scaled, train[target_feature], "gd")
sgd_losses, sgd_times, sgd_iters = run_experiment(X_train_scaled, train[target_feature], "sgd")
```

### Задача 6.

В этой задаче вам предлагается реализовать регрессию Хьюбера, а также применить ее к данным с выбросами. Для начала реализуйте класс по шаблону снизу. Обратите внимание, что класс `HuberRegression` &mdash; наследник класса `BaseEstimator`, это с легкостью позволит использовать наш класс в различных пайплайнах библиотеки `sklearn`.

**1.** Задача оптимизации для регрессии Хьюбера выглядит следующим образом:
$$\sum_{i=1}^n R\left(Y_i - x_i^T\theta\right) \rightarrow \min_\theta,$$
где $R(x)$ &mdash; функция потерь Хьюбера, определяемая как
$$R(x) = \frac{x^2}{2} I\left\{|x| < c\right\} + c \left(|x| - \frac{c}{2}\right)I\left\{|x| > c\right\}.$$

Выпишите формулы для градиентного и стохастического градиентного спусков. В чем польза такой функции потерь?

Градиент функции потерь:
$$
\nabla L(x) =
\begin{cases} 
- x, & \text{если } |x| \leq c \\
- c \cdot \text{sign}(x), & \text{если } |x| > c
\end{cases}
$$

Градиентный спуск:
$$
\theta^{(t+1)} = \theta^{(t)} - \eta \cdot \frac{1}{N} \sum_{i=1}^{N} \nabla L_(error_i) X_i
$$

Стохастический градиентный спуск:
$$
\theta^{(t+1)} = \theta^{(t)} - \eta \cdot \frac{1}{|B|} \sum_{i \in B} \nabla L_(error_i) X_i
$$

где 
$$error_i = y - \hat{y}$$

**2.** Реализуем теперь класс. **При реализации класса запрещено пользоваться ИИ-инструментами.**

```python
import numpy as np
from sklearn.base import BaseEstimator

class HuberRegression(BaseEstimator):
    """Класс, реализующий линейную регрессию с функцией потерь Хьюбера."""

    def __init__(self, c: float = 1.0, fit_intercept: bool = True, max_iter: int = 1000, learning_rate: float = 0.5):
        """Инициализирует модель.

        Параметры:
        c (float): Константа из функции потерь Хьюбера. Определяет порог, при котором функция потерь переходит
                   из квадратичной в линейную.
        fit_intercept (bool): Добавлять ли константный признак (свободный коэффициент).
        max_iter (int): Максимальное число итераций оптимизации.
        learning_rate (float): Скорость обучения, на которую домножается градиент при обновлении весов.
        """
        self.c = c  
        self.fit_intercept = fit_intercept  
        self.max_iter = max_iter  
        self.learning_rate = learning_rate  

    def _huber_gradient(self, y_true, y_pred, X):
        """Вычисляет градиент функции потерь Хьюбера.

        Параметры:
        y_true (np.ndarray): Истинные значения целевой переменной.
        y_pred (np.ndarray): Предсказанные значения целевой переменной.
        X (np.ndarray): Матрица признаков.

        Возвращает:
        np.ndarray: Градиент функции потерь по весам модели.
        """
        error = y_true - y_pred  
        mask = np.abs(error) <= self.c  
        gradient = np.where(mask, -error, -self.c * np.sign(error))  
        return np.dot(X.T, gradient) / len(y_true)  
    
    def fit(self, X: np.ndarray, y: np.ndarray) -> "HuberRegression":
        """Обучает модель.

        Параметры:
        X (np.ndarray): Матрица признаков.
        y (np.ndarray): Вектор целевой переменной.

        Возвращает:
        HuberRegression: Обученная модель.
        """
        # Проверка на совпадение количества строк в X и y
        if X.shape[0] != y.shape[0]:
            raise ValueError("Количество строк в X и y должно совпадать")

        # Добавление свободного коэффициента, если требуется
        if self.fit_intercept:
            X = np.c_[np.ones(X.shape[0]), X] 

        # Инициализация весов случайными значениями
        self.coef_ = np.random.randn(X.shape[1])

        # Градиентный спуск
        for i in range(self.max_iter):
            y_pred = X @ self.coef_  
            gradient = self._huber_gradient(y, y_pred, X)  
            self.coef_ -= self.learning_rate * gradient  

            if np.linalg.norm(gradient) < 1e-6:
                break

        self.n_iter_ = i 

        # Разделение коэффициентов на свободный член и остальные веса
        if self.fit_intercept:
            self.intercept_ = self.coef_[0]  
            self.coef_ = self.coef_[1:]  
        else:
            self.intercept_ = 0.0  

        return self

    def predict(self, X: np.ndarray) -> np.ndarray:
        """Делает предсказание на новых данных.

        Параметры:
        X (np.ndarray): Матрица признаков.

        Возвращает:
        np.ndarray: Вектор предсказанных значений.
        """
        # Добавление свободного коэффициента, если требуется
        if self.fit_intercept:
            X = np.c_[np.ones(X.shape[0]), X]

        return X @ np.append(self.intercept_, self.coef_)
```

### Задача 7.


Рассмотрим модель одномерной регрессии $y(x) = \theta x$, где $x \in \mathbb{R}$ &mdash; одномерный признак, $y \in \mathbb{R}$ &mdash; целевой признак, $\theta \in \mathbb{R}$ &mdash; неизвестный параметр. Имеется выборка размера $n$, полученная по правилу
	$$Y_i = \theta x_i + \varepsilon_i,\ \ \ i=1,...,n,$$
где $\varepsilon_i$ &mdash; случайная ошибка измерений.

Предложите точный алгоритм поиска оценки параметра $\theta$ методом наименьших модулей, то есть $$\sum_{i=1}^n \left|Y_i - \theta x_i\right| \to \min_\theta,$$ работающий за время $O(n \log n)$. Приведите его описание и теоретическое обоснование. Реализация в коде не требуется.

**Решение:**

Мы хотим минимизировать функцию потерь:
$$
\sum_{i=1}^n |Y_i - \theta x_i| \to \min_\theta.
$$
Это эквивалентно минимизации:
$$
\sum_{i=1}^n w_i |r_i - \theta|,
$$
где:
- $r_i = Y_i / x_i$ (если $x_i \neq 0$),
- $w_i = |x_i|$ (веса).

Минимум этой функции достигается в точке **взвешенной медианы** значений $r_i$ с весами $w_i$.

**Алгоритм**

2. **Вычисление $r_i$ и весов $w_i$**:
   Для каждой точки вычислим $r_i = Y_i / x_i$ и $w_i = |x_i|$. Сформируем из них пары $(r_i, w_i)$ и создадим массив. Это сработает за $O(n)$

3. **Сортировка**:
   Отсортируем массив по возрастанию $r_i$. Это займёт $O(n \log n)$.

4. **Вычисление суммарного веса**:
   Найдем суммарный вес $W = \sum_{i=1}^n w_i$. (Это модно было сделать на первой итерации, тоже за $O(n)$)

5. **Поиск взвешенной медианы**:
   Пройдем по отсортированному массиву, считая сумму $w_i$ 
   Если нашелся $k$, при котором накопленная сумма весов впервые превышает $W/2$, т.е.
     $$
     \sum_{i=1}^{k-1} w_i < \frac{W}{2} \quad \text{и} \quad \sum_{i=1}^k w_i \geq \frac{W}{2}.
     $$
   То мы останавливаем наш проход и возвращаем $r_k$, этот этап работает за $O(n)$

**Коректность** :  Функция потерь $\sum w_i |r_i - \theta|$ — это взвешенная сумма модулей отклонений.  
Минимум такой функции достигается в точке, где сумма весов слева и справа от $\theta$ примерно равна. Это и есть определение **взвешенной медианы**.
